```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
merge_df=pd.read_csv('/content/drive/MyDrive/DL_merged_dataset.csv')
```

PIE CHART(MERGED DATASET)

```python
merge_df['rating'].value_counts().sort_index().plot.pie(autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title("Ratings Proportion")
plt.ylabel('')
plt.show()
```

BAR CHART(MERGED DATASET)

```python
x=merge_df['rating'].value_counts().index
y=merge_df['rating'].value_counts().values
sns.barplot(x=x, y=y)
plt.title("BAR PLOT-RATINGS COUNT")
plt.xlabel("Ratings")
plt.ylabel("Count")
plt.show()
```

HORIZONTAL BAR CHART(MERGED DATASET)

```python
plt.barh(x,y)
plt.title("HORIZONTAL BAR PLOT-RATINGS COUNT")
plt.xlabel("Count")
plt.ylabel("Ratings")
plt.show()
```
