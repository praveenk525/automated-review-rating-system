CREATING A BALANCED DATASET

```python
import pandas as pd
bal_df = (dl_merge_df.groupby('rating', group_keys=False)
  .apply(lambda x: x.sample(n=150000, random_state=42)))

remain_df=dl_merge_df.drop(bal_df.index)


bal_df= bal_df.reset_index(drop=True)
remain_df=remain_df.reset_index(drop=True)
```

```python
bal_df['rating'].value_counts()
```

```python
remain_df['rating'].value_counts()
```

```python
remain_df = remain_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Desired sample sizes based on % targets
target_counts = {
    5: 858540,
    4: 715450,
    3: 429270,
    2: 286180,
    1: 572360

}

# Apply sampling
sampled = []

for rating, count in target_counts.items():
    group = remain_df[remain_df['rating'] == rating]
    sampled.append(group.iloc[:count])  # get top N

final_df = pd.concat(sampled).reset_index(drop=True)

# Check result
print(final_df['rating'].value_counts(normalize=True) * 100)  # % distribution
```
